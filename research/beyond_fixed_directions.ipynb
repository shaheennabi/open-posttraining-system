### Beyond Fixed Directions: Understanding the Representational Geometry of Reasoning and Memorization in Large Language Models

---

#### Core Scientific Question

**How are reasoning and memorization represented inside large language models, and how do these representations evolve during reinforcement learning?**

Rather than assuming reasoning and memorization can be represented by a single fixed direction, this work systematically studies:

- Their internal geometry
- Their evolution during reinforcement learning
- Their ability to generalize across tasks
- Their internal mechanistic composition

> **DiRL is an applied motivation, not the subject of the paper.**

---

#### Research Hypothesis

Reasoning and memorization may **not** be adequately represented by a single fixed linear direction.

Instead, they may emerge as **adaptive, evolving, multi-component latent representations** whose geometry changes during reinforcement learning and varies across tasks.

This is the hypothesis that the experiments are designed to test—not a conclusion assumed beforehand.

---

#### Research Objectives

1. Characterize the geometry of reasoning and memorization.
2. Study how these representations evolve during reinforcement learning.
3. Understand whether they generalize across different reasoning tasks.
4. Identify the latent components that compose reasoning and memorization.

````python
### the tokenizer also adds the formatted assistant/special tokens
### before generation
###
### so for our primary experiment we take the representation
### of the LAST FORMATTED PROMPT TOKEN


def get100(problems, device, prompt_key="problem"):  ## prompt_key as --> if you have memorization data pass question, otherwise problem for reasoning dataset

    ## store hidden representations for all 100 problems
    all_hidden_layers = []

    model.eval()  ## get the model in the eval mode.. it will add ..--->  disabling layers like dropout, and using running batchnorm statistics from training

    with torch.no_grad():

        ## go problem-by-problem
        for i in problems:
            
            ## get current problem
            prompt = i[prompt_key]  ## this should work for both the reasoning and memoriation data


            ## tokenize the complete formatted prompt
            ## then convert to tensor
            ##
            ## shape --> (1, seq_len)
            input_ids = torch.tensor(tokenizer.encode(prompt), dtype=torch.long, device=device).unsqueeze(0)


            ## forward pass
            ##
            ## hidden_layers shape:
            ## (1, n_layers, emb_dim)
            _ , hidden_layers = model(input_ids, cache=None)


            ## remove batch dimension
            ##
            ## (1, n_layers, emb_dim)
            ##          ↓
            ## (n_layers, emb_dim)
            hidden_layers = hidden_layers.squeeze(0)


            ## move representations to CPU
            ## and store current problem
            all_hidden_layers.append(hidden_layers.detach().cpu())


    ## after processing all 100 problems:
    ##
    ## list:
    ##
    ## problem 1   --> (n_layers, emb_dim)
    ## problem 2   --> (n_layers, emb_dim)
    ## ...
    ## problem 100 --> (n_layers, emb_dim)
    ##
    ## stack them:
    ##
    ## (100, n_layers, emb_dim)
    all_hidden_layers = torch.stack(all_hidden_layers, dim=0)


    return all_hidden_layers
````

## Length-matched MATH-500 vs PopQA control experiment

This section controls the strong prompt-token-length confound found in the original MATH-500 vs PopQA comparison. It does not modify question content. Each helper is defined directly in this notebook so the experiment is self-contained for Lightning AI Studio.

Run the first runner to build/verify the matched subset. Run the GPU runner only after the length-only baseline is acceptable. The extraction uses the existing custom `model` and `tokenizer`; it does not replace them with Hugging Face implementations.


````python
# Length-matched control experiment imports and constants
from copy import deepcopy
from pathlib import Path
import json
import math

import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt

from sklearn.decomposition import PCA
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, balanced_accuracy_score, roc_auc_score
from sklearn.model_selection import StratifiedKFold


REASONING_LABEL = 1
MEMORIZATION_LABEL = 0
````

````python
def get_prompt_text(example, prompt_key=None):
    """Return the text field used by the existing extraction pipeline."""
    if prompt_key is not None:
        return example[prompt_key]
    if "problem" in example:
        return example["problem"]
    if "question" in example:
        return example["question"]
    raise KeyError("Example must contain either 'problem' or 'question', or pass prompt_key.")
````

````python
def token_length_after_pipeline(example, tokenizer, prompt_key=None):
    """Measure length after the same tokenizer.encode(prompt) call used for hidden extraction."""
    prompt = get_prompt_text(example, prompt_key=prompt_key)
    return len(tokenizer.encode(prompt))
````

````python
def _summary_stats(values):
    values = np.asarray(values, dtype=float)
    return {
        "count": int(values.size),
        "min": float(np.min(values)),
        "max": float(np.max(values)),
        "mean": float(np.mean(values)),
        "median": float(np.median(values)),
        "std": float(np.std(values, ddof=1)) if values.size > 1 else 0.0,
        "p25": float(np.percentile(values, 25)),
        "p50": float(np.percentile(values, 50)),
        "p75": float(np.percentile(values, 75)),
    }
````

````python
def _print_stats(name, values):
    stats = _summary_stats(values)
    print(f"\n{name}")
    print("-" * len(name))
    for key in ["count", "min", "max", "mean", "median", "std", "p25", "p50", "p75"]:
        value = stats[key]
        if key == "count":
            print(f"{key:>8}: {value}")
        else:
            print(f"{key:>8}: {value:.3f}")
    return stats
````

````python
def analyze_token_lengths(
    reasoning_examples,
    memorization_examples,
    tokenizer,
    reasoning_prompt_key="problem",
    memorization_prompt_key="question",
    plot=True,
):
    """Tokenize both original datasets and summarize their prompt-token lengths."""
    records = []

    for index, example in enumerate(reasoning_examples):
        token_length = token_length_after_pipeline(example, tokenizer, reasoning_prompt_key)
        records.append(
            {
                "index": index,
                "id": example.get("id", index),
                "dataset": example.get("dataset", "MATH-500"),
                "label": "reasoning",
                "label_id": REASONING_LABEL,
                "token_length": token_length,
            }
        )

    for index, example in enumerate(memorization_examples):
        token_length = token_length_after_pipeline(example, tokenizer, memorization_prompt_key)
        records.append(
            {
                "index": index,
                "id": example.get("id", index),
                "dataset": example.get("dataset", "PopQA"),
                "label": "memorization",
                "label_id": MEMORIZATION_LABEL,
                "token_length": token_length,
            }
        )

    length_df = pd.DataFrame(records)
    reasoning_lengths = length_df.loc[length_df["label"] == "reasoning", "token_length"].to_numpy()
    memorization_lengths = length_df.loc[length_df["label"] == "memorization", "token_length"].to_numpy()

    print("Original token-length distributions")
    print("=" * 44)
    reasoning_stats = _print_stats("Reasoning", reasoning_lengths)
    memorization_stats = _print_stats("Memorization", memorization_lengths)

    if plot:
        plt.figure(figsize=(9, 5))
        plt.hist(reasoning_lengths, bins=20, alpha=0.60, label="Reasoning")
        plt.hist(memorization_lengths, bins=20, alpha=0.60, label="Memorization")
        plt.xlabel("Prompt token length")
        plt.ylabel("Example count")
        plt.title("Before length matching")
        plt.legend()
        plt.tight_layout()
        plt.show()

    return {
        "length_df": length_df,
        "reasoning_stats": reasoning_stats,
        "memorization_stats": memorization_stats,
    }
````

````python
def _maximum_bipartite_matching(reasoning_lengths, memorization_lengths, max_length_difference):
    """Maximum-cardinality one-to-one matching within the length tolerance."""
    edges = []
    for r_idx, r_len in enumerate(reasoning_lengths):
        possible = []
        for m_idx, m_len in enumerate(memorization_lengths):
            diff = abs(int(r_len) - int(m_len))
            if diff <= max_length_difference:
                possible.append((diff, m_idx))
        edges.append([m_idx for _, m_idx in sorted(possible)])

    match_mem_to_reason = {}

    def try_assign(r_idx, seen):
        for m_idx in edges[r_idx]:
            if m_idx in seen:
                continue
            seen.add(m_idx)
            if m_idx not in match_mem_to_reason or try_assign(match_mem_to_reason[m_idx], seen):
                match_mem_to_reason[m_idx] = r_idx
                return True
        return False

    for r_idx in np.argsort(reasoning_lengths):
        try_assign(int(r_idx), set())

    pairs = [(r_idx, m_idx) for m_idx, r_idx in match_mem_to_reason.items()]
    pairs.sort(key=lambda pair: (reasoning_lengths[pair[0]], pair[0]))
    return pairs
````

````python
def match_by_token_length(
    reasoning_examples,
    memorization_examples,
    tokenizer,
    max_length_difference=5,
    reasoning_prompt_key="problem",
    memorization_prompt_key="question",
):
    """Create a new one-to-one length-matched dataset without modifying the originals."""
    reasoning_lengths = np.array(
        [token_length_after_pipeline(example, tokenizer, reasoning_prompt_key) for example in reasoning_examples]
    )
    memorization_lengths = np.array(
        [token_length_after_pipeline(example, tokenizer, memorization_prompt_key) for example in memorization_examples]
    )

    pairs = _maximum_bipartite_matching(reasoning_lengths, memorization_lengths, max_length_difference)
    matched_reasoning = []
    matched_memorization = []

    for match_id, (r_idx, m_idx) in enumerate(pairs, start=1):
        r_example = deepcopy(reasoning_examples[r_idx])
        m_example = deepcopy(memorization_examples[m_idx])
        r_len = int(reasoning_lengths[r_idx])
        m_len = int(memorization_lengths[m_idx])
        diff = abs(r_len - m_len)

        r_example.update(
            {
                "dataset": r_example.get("dataset", "MATH-500"),
                "label": "reasoning",
                "label_id": REASONING_LABEL,
                "token_length": r_len,
                "match_id": match_id,
                "matched_with_label": "memorization",
                "matched_with_original_index": int(m_idx),
                "matched_with_id": memorization_examples[m_idx].get("id", int(m_idx)),
                "length_difference": int(diff),
                "max_length_difference": int(max_length_difference),
            }
        )
        m_example.update(
            {
                "dataset": m_example.get("dataset", "PopQA"),
                "label": "memorization",
                "label_id": MEMORIZATION_LABEL,
                "token_length": m_len,
                "match_id": match_id,
                "matched_with_label": "reasoning",
                "matched_with_original_index": int(r_idx),
                "matched_with_id": reasoning_examples[r_idx].get("id", int(r_idx)),
                "length_difference": int(diff),
                "max_length_difference": int(max_length_difference),
            }
        )
        matched_reasoning.append(r_example)
        matched_memorization.append(m_example)

    print("\nLength matching result")
    print("=" * 44)
    print("max_length_difference       :", max_length_difference)
    print("original reasoning count    :", len(reasoning_examples))
    print("original memorization count :", len(memorization_examples))
    print("matched reasoning count     :", len(matched_reasoning))
    print("matched memorization count  :", len(matched_memorization))
    print("discarded reasoning count   :", len(reasoning_examples) - len(matched_reasoning))
    print("discarded memorization count:", len(memorization_examples) - len(matched_memorization))

    if len(matched_reasoning) < min(len(reasoning_examples), len(memorization_examples)):
        print("WARNING: full 100-vs-100 matching was not possible at this tolerance.")

    return {
        "matched_reasoning": matched_reasoning,
        "matched_memorization": matched_memorization,
        "matched_dataset": matched_reasoning + matched_memorization,
        "pairs": pairs,
        "reasoning_lengths": reasoning_lengths,
        "memorization_lengths": memorization_lengths,
    }
````

````python
def _as_metric_summary(values):
    values = np.asarray(values, dtype=float)
    return {"mean": float(values.mean()), "std": float(values.std()), "values": values.tolist()}
````

````python
def _ensure_5_fold_possible(y):
    counts = np.bincount(np.asarray(y, dtype=int))
    min_class_count = int(counts[counts > 0].min())
    if min_class_count < 5:
        raise ValueError(f"Need at least 5 examples per class for 5-fold CV, got {min_class_count}.")
````

````python
def run_linear_probe(X, y, title="Linear probe", random_state=42):
    """Run the same 5-fold LogisticRegression probe used in the original notebook."""
    X = np.asarray(X, dtype=np.float32)
    y = np.asarray(y, dtype=int)
    _ensure_5_fold_possible(y)

    cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=random_state)
    accuracies = []
    balanced_accuracies = []
    aurocs = []
    fold_rows = []

    print("\n" + title)
    print("=" * len(title))
    for fold, (train_idx, test_idx) in enumerate(cv.split(X, y), start=1):
        probe = LogisticRegression(C=1.0, max_iter=5000)
        probe.fit(X[train_idx], y[train_idx])
        y_pred = probe.predict(X[test_idx])
        y_prob = probe.predict_proba(X[test_idx])[:, 1]

        accuracy = accuracy_score(y[test_idx], y_pred)
        balanced_accuracy = balanced_accuracy_score(y[test_idx], y_pred)
        auroc = roc_auc_score(y[test_idx], y_prob)

        accuracies.append(accuracy)
        balanced_accuracies.append(balanced_accuracy)
        aurocs.append(auroc)
        fold_rows.append(
            {
                "fold": fold,
                "accuracy": float(accuracy),
                "balanced_accuracy": float(balanced_accuracy),
                "auroc": float(auroc),
            }
        )
        print(f"Fold {fold}: accuracy={accuracy:.4f}, balanced_accuracy={balanced_accuracy:.4f}, auroc={auroc:.4f}")

    results = {
        "folds": fold_rows,
        "accuracy": _as_metric_summary(accuracies),
        "balanced_accuracy": _as_metric_summary(balanced_accuracies),
        "auroc": _as_metric_summary(aurocs),
    }
    print("\nSummary")
    print("-" * 20)
    print(f"Accuracy          : {results['accuracy']['mean']:.4f} +/- {results['accuracy']['std']:.4f}")
    print(f"Balanced Accuracy : {results['balanced_accuracy']['mean']:.4f} +/- {results['balanced_accuracy']['std']:.4f}")
    print(f"AUROC             : {results['auroc']['mean']:.4f} +/- {results['auroc']['std']:.4f}")
    return results
````

````python
def length_only_baseline(matched_reasoning, matched_memorization, warning_accuracy=0.80, warning_auroc=0.80):
    """Classify reasoning vs memorization using only prompt token length."""
    token_lengths = [x["token_length"] for x in matched_reasoning] + [x["token_length"] for x in matched_memorization]
    X_length = np.asarray(token_lengths, dtype=np.float32).reshape(-1, 1)
    y = np.asarray([REASONING_LABEL] * len(matched_reasoning) + [MEMORIZATION_LABEL] * len(matched_memorization), dtype=int)
    results = run_linear_probe(X_length, y, title="Length-only baseline")
    if results["accuracy"]["mean"] >= warning_accuracy or results["auroc"]["mean"] >= warning_auroc:
        print("\nWARNING: token length is still strongly predictive on the matched dataset.")
        print("Do not interpret hidden-state probe results until this confound is reduced.")
    return results
````

````python
def run_pca(X, y, title="Final-layer PCA", plot=True):
    """Fit PCA without using labels, then use labels only for plotting."""
    X = np.asarray(X, dtype=np.float32)
    y = np.asarray(y, dtype=int)
    pca = PCA(n_components=2)
    X_pca = pca.fit_transform(X)
    explained = pca.explained_variance_ratio_
    print("\n" + title)
    print("=" * len(title))
    print(f"PC1 explained variance ratio     : {explained[0]:.4f}")
    print(f"PC2 explained variance ratio     : {explained[1]:.4f}")
    print(f"PC1 + PC2 explained variance     : {explained.sum():.4f}")
    if plot:
        plt.figure(figsize=(8, 6))
        plt.scatter(X_pca[y == REASONING_LABEL, 0], X_pca[y == REASONING_LABEL, 1], label="Reasoning", alpha=0.75)
        plt.scatter(
            X_pca[y == MEMORIZATION_LABEL, 0],
            X_pca[y == MEMORIZATION_LABEL, 1],
            label="Memorization",
            alpha=0.75,
        )
        plt.xlabel("PC1")
        plt.ylabel("PC2")
        plt.title(title)
        plt.legend()
        plt.tight_layout()
        plt.show()
    return {
        "pca": pca,
        "X_pca": X_pca,
        "pc1_variance": float(explained[0]),
        "pc2_variance": float(explained[1]),
        "pc1_pc2_variance": float(explained.sum()),
    }
````

````python
def run_random_label_baseline(X, y, random_state=123):
    """Shuffle labels while keeping X unchanged, then rerun the same 5-fold probe."""
    rng = np.random.default_rng(random_state)
    shuffled_y = np.asarray(y, dtype=int).copy()
    rng.shuffle(shuffled_y)
    return run_linear_probe(X, shuffled_y, title="Random-label control", random_state=42)
````

````python
def extract_matched_hidden_states(
    matched_reasoning,
    matched_memorization,
    model,
    tokenizer,
    device=None,
    reasoning_prompt_key="problem",
    memorization_prompt_key="question",
):
    """Extract [N, 28, 1024] hidden states for the matched dataset using the existing Qwen3 model."""
    if device is None:
        device = "cuda" if torch.cuda.is_available() else "cpu"
    model.to(device)
    model.eval()

    def extract_group(examples, prompt_key):
        all_hidden_layers = []
        with torch.no_grad():
            for example in examples:
                prompt = get_prompt_text(example, prompt_key=prompt_key)
                input_ids = torch.tensor(tokenizer.encode(prompt), dtype=torch.long, device=device).unsqueeze(0)
                _, hidden_layers = model(input_ids, cache=None)
                all_hidden_layers.append(hidden_layers.squeeze(0).detach().cpu())
        return torch.stack(all_hidden_layers, dim=0)

    reasoning_hidden = extract_group(matched_reasoning, reasoning_prompt_key)
    memorization_hidden = extract_group(matched_memorization, memorization_prompt_key)
    print("Matched hidden-state extraction complete")
    print("reasoning_hidden shape    :", tuple(reasoning_hidden.shape))
    print("memorization_hidden shape :", tuple(memorization_hidden.shape))
    return {
        "reasoning_hidden_states": reasoning_hidden,
        "memorization_hidden_states": memorization_hidden,
        "matched_reasoning_metadata": matched_reasoning,
        "matched_memorization_metadata": matched_memorization,
        "extraction_config": {
            "model": "Qwen3-0.6B custom implementation",
            "tokenizer": "custom Qwen3 tokenizer",
            "representation": "last_formatted_prompt_token",
            "position": -1,
            "boundary": "post_transformer_block_pre_final_norm",
            "use_kv_cache": False,
            "generation": False,
            "num_reasoning_samples": int(reasoning_hidden.shape[0]),
            "num_memorization_samples": int(memorization_hidden.shape[0]),
            "n_layers": int(reasoning_hidden.shape[1]),
            "hidden_dim": int(reasoning_hidden.shape[2]),
        },
    }
````

````python
def final_layer_matrix(extraction_result):
    """Return final-layer X and labels from extracted matched hidden states."""
    reasoning_X = extraction_result["reasoning_hidden_states"][:, -1, :]
    memorization_X = extraction_result["memorization_hidden_states"][:, -1, :]
    X = torch.cat([reasoning_X, memorization_X], dim=0).float().cpu().numpy()
    y = np.asarray([REASONING_LABEL] * reasoning_X.shape[0] + [MEMORIZATION_LABEL] * memorization_X.shape[0], dtype=int)
    return X, y, reasoning_X, memorization_X
````

````python
# Prompt normalization and token-length helpers

def expanded_format_prompt(question):
    # One common template for MATH-500, GSM8K, PopQA, and TriviaQA.
    return (
        "Answer the following question.\n\n"
        f"Question: {question}\n\n"
        "Answer:"
    )


def expanded_token_length(question, tokenizer):
    formatted_prompt = expanded_format_prompt(question)
    return len(tokenizer.encode(formatted_prompt))


def expanded_add_prompt_and_lengths(examples, tokenizer):
    enriched = []
    for ex in examples:
        item = deepcopy(ex)
        item["formatted_prompt"] = expanded_format_prompt(item["question"])
        item["token_length"] = len(tokenizer.encode(item["formatted_prompt"]))
        enriched.append(item)
    return enriched
````

````python
# Dataset loading and schema unification

def expanded_inspect_dependencies():
    print("HF datasets available:", HF_DATASETS_AVAILABLE)
    print("Expected existing local files:")
    for path in [Path("math500.json"), Path("popQA.json")]:
        print(f"  {path}: exists={path.exists()}")


def expanded_first_existing_key(record, keys, default=None):
    for key in keys:
        if key in record and record[key] is not None:
            return record[key]
    return default


def expanded_load_local_math500(path="math500.json"):
    data = expanded_load_json(path)
    unified = []
    for i, row in enumerate(data):
        question = expanded_first_existing_key(row, ["problem", "question"])
        answer = expanded_first_existing_key(row, ["answer", "solution", "final_answer"])
        if question is None:
            continue
        unified.append({
            "id": expanded_first_existing_key(row, ["id", "unique_id"], i),
            "dataset": "math500",
            "task_type": REASONING_TASK,
            "question": str(question),
            "answer": answer,
            "source_metadata": deepcopy(row),
        })
    print("Loaded MATH-500 local examples:", len(unified))
    if unified:
        print("MATH-500 example keys:", sorted(unified[0].keys()))
    return unified


def expanded_load_local_popqa(path="popQA.json"):
    data = expanded_load_json(path)
    unified = []
    for i, row in enumerate(data):
        question = expanded_first_existing_key(row, ["question", "query"])
        answer = expanded_first_existing_key(row, ["answer", "answers", "possible_answers", "obj"])
        if question is None:
            continue
        unified.append({
            "id": expanded_first_existing_key(row, ["id", "subj", "uid"], i),
            "dataset": "popqa",
            "task_type": FACTUAL_RECALL_TASK,
            "question": str(question),
            "answer": answer,
            "source_metadata": deepcopy(row),
        })
    print("Loaded PopQA local examples:", len(unified))
    if unified:
        print("PopQA example keys:", sorted(unified[0].keys()))
    return unified


def expanded_load_gsm8k(split="train"):
    if not HF_DATASETS_AVAILABLE:
        raise ImportError("datasets is not installed. Install/load it in Lightning AI before running this cell.")
    ds = load_dataset("openai/gsm8k", "main", split=split)
    print("GSM8K schema:", ds.features)
    unified = []
    for i, row in enumerate(ds):
        question = row.get("question")
        answer = row.get("answer")
        if question is None:
            continue
        unified.append({
            "id": f"gsm8k_{split}_{i}",
            "dataset": "gsm8k",
            "task_type": REASONING_TASK,
            "question": str(question),
            "answer": answer,
            "source_metadata": dict(row),
        })
    print("Loaded GSM8K examples:", len(unified))
    return unified


def expanded_load_triviaqa(split="train"):
    if not HF_DATASETS_AVAILABLE:
        raise ImportError("datasets is not installed. Install/load it in Lightning AI before running this cell.")
    try:
        ds = load_dataset("trivia_qa", "rc.nocontext", split=split)
        trivia_config = "rc.nocontext"
    except Exception:
        ds = load_dataset("trivia_qa", "unfiltered.nocontext", split=split)
        trivia_config = "unfiltered.nocontext"
    print("TriviaQA config:", trivia_config)
    print("TriviaQA schema:", ds.features)
    unified = []
    for i, row in enumerate(ds):
        question = row.get("question")
        answer = row.get("answer")
        if isinstance(answer, dict):
            answer_text = answer.get("value") or answer.get("aliases")
        else:
            answer_text = answer
        if question is None:
            continue
        unified.append({
            "id": row.get("question_id", f"triviaqa_{split}_{i}"),
            "dataset": "triviaqa",
            "task_type": FACTUAL_RECALL_TASK,
            "question": str(question),
            "answer": answer_text,
            "source_metadata": dict(row),
        })
    print("Loaded TriviaQA examples:", len(unified))
    return unified


def expanded_load_all_datasets(paths, gsm8k_split="train", triviaqa_split="train"):
    expanded_inspect_dependencies()
    math500 = expanded_load_local_math500("math500.json")
    popqa = expanded_load_local_popqa("popQA.json")
    gsm8k = expanded_load_gsm8k(split=gsm8k_split)
    triviaqa = expanded_load_triviaqa(split=triviaqa_split)
    all_examples = math500 + gsm8k + popqa + triviaqa
    expanded_save_json(math500, paths["unified"] / "math500_unified.json")
    expanded_save_json(gsm8k, paths["unified"] / "gsm8k_unified.json")
    expanded_save_json(popqa, paths["unified"] / "popqa_unified.json")
    expanded_save_json(triviaqa, paths["unified"] / "triviaqa_unified.json")
    expanded_save_json(all_examples, paths["unified"] / "all_unified.json")
    print("Unified total examples:", len(all_examples))
    print("Counts by dataset:", Counter(ex["dataset"] for ex in all_examples))
    print("Counts by task_type:", Counter(ex["task_type"] for ex in all_examples))
    return all_examples
````

````python
# Deduplication and token-length analysis

def expanded_normalize_question_for_dedup(question):
    return re.sub(r"\s+", " ", str(question).strip().lower())


def expanded_deduplicate_examples(examples):
    seen_questions = set()
    seen_dataset_ids = set()
    deduped = []
    removed = []
    for ex in examples:
        q_key = expanded_normalize_question_for_dedup(ex["question"])
        id_key = (ex["dataset"], str(ex["id"]))
        reasons = []
        if q_key in seen_questions:
            reasons.append("duplicate_question")
        if id_key in seen_dataset_ids:
            reasons.append("duplicate_dataset_id")
        if reasons:
            removed.append({
                "id": ex.get("id"),
                "dataset": ex.get("dataset"),
                "task_type": ex.get("task_type"),
                "question": ex.get("question"),
                "remove_reasons": reasons,
            })
            continue
        seen_questions.add(q_key)
        seen_dataset_ids.add(id_key)
        deduped.append(ex)
    print("Deduplication")
    print("  input examples  :", len(examples))
    print("  kept examples   :", len(deduped))
    print("  removed examples:", len(removed))
    print("  removal reasons :", Counter(reason for row in removed for reason in row["remove_reasons"]))
    return deduped, removed


def expanded_analyze_lengths(examples, title="Token length statistics", plot=True, save_dir=None):
    rows = []
    for ex in examples:
        rows.append({
            "id": ex["id"],
            "dataset": ex["dataset"],
            "task_type": ex["task_type"],
            "token_length": ex["token_length"],
        })
    df = pd.DataFrame(rows)
    by_dataset = {k: expanded_stats(v["token_length"].to_numpy()) for k, v in df.groupby("dataset")}
    by_task = {k: expanded_stats(v["token_length"].to_numpy()) for k, v in df.groupby("task_type")}
    dataset_df = expanded_print_stats_table(title + " by dataset", by_dataset)
    task_df = expanded_print_stats_table(title + " by task type", by_task)
    if plot:
        fig, axes = plt.subplots(1, 2, figsize=(14, 5))
        for dataset_name, sub in df.groupby("dataset"):
            axes[0].hist(sub["token_length"], bins=30, alpha=0.5, label=dataset_name)
        axes[0].set_title(title + " by dataset")
        axes[0].set_xlabel("Token length")
        axes[0].set_ylabel("Count")
        axes[0].legend()
        for task_type, sub in df.groupby("task_type"):
            axes[1].hist(sub["token_length"], bins=30, alpha=0.5, label=task_type)
        axes[1].set_title(title + " by task type")
        axes[1].set_xlabel("Token length")
        axes[1].legend()
        plt.tight_layout()
        if save_dir is not None:
            save_dir = Path(save_dir)
            save_dir.mkdir(parents=True, exist_ok=True)
            fig.savefig(save_dir / (title.lower().replace(" ", "_") + ".png"), dpi=200)
        plt.show()
    return {
        "table": df,
        "by_dataset": by_dataset,
        "by_task": by_task,
        "dataset_df": dataset_df,
        "task_df": task_df,
    }
````

````python
# Distributional length matching

def expanded_length_bin(token_length, bin_width=EXPANDED_BIN_WIDTH):
    return int(token_length // bin_width) * bin_width


def expanded_distributional_length_match(
    examples,
    target_per_class=EXPANDED_TARGET_PER_CLASS,
    bin_width=EXPANDED_BIN_WIDTH,
    random_seed=EXPANDED_RANDOM_SEED,
):
    rng = random.Random(random_seed)
    by_task_bin = defaultdict(list)
    for ex in examples:
        task = ex["task_type"]
        bin_start = expanded_length_bin(ex["token_length"], bin_width=bin_width)
        by_task_bin[(task, bin_start)].append(ex)
    all_bins = sorted({bin_start for _, bin_start in by_task_bin.keys()})
    capacity_by_bin = {}
    for bin_start in all_bins:
        r_count = len(by_task_bin[(REASONING_TASK, bin_start)])
        f_count = len(by_task_bin[(FACTUAL_RECALL_TASK, bin_start)])
        capacity_by_bin[bin_start] = min(r_count, f_count)
    total_capacity_per_class = sum(capacity_by_bin.values())
    actual_per_class = min(target_per_class, total_capacity_per_class)
    if actual_per_class == 0:
        raise ValueError("No overlapping token-length bins between reasoning and factual_recall.")
    selected_per_bin = {}
    remaining = actual_per_class
    for bin_start in all_bins:
        take = min(capacity_by_bin[bin_start], remaining)
        selected_per_bin[bin_start] = take
        remaining -= take
        if remaining == 0:
            break
    matched = []
    selected_by_task = {REASONING_TASK: [], FACTUAL_RECALL_TASK: []}
    for bin_start, take in selected_per_bin.items():
        if take <= 0:
            continue
        for task in [REASONING_TASK, FACTUAL_RECALL_TASK]:
            candidates = list(by_task_bin[(task, bin_start)])
            rng.shuffle(candidates)
            chosen = candidates[:take]
            for ex in chosen:
                item = deepcopy(ex)
                item["length_bin_start"] = bin_start
                item["length_bin_end"] = bin_start + bin_width - 1
                item["matching_method"] = "distributional_equal_bin_sampling"
                item["matching_random_seed"] = random_seed
                item["matching_bin_width"] = bin_width
                item["requested_target_per_class"] = target_per_class
                item["actual_target_per_class"] = actual_per_class
                selected_by_task[task].append(item)
                matched.append(item)
    rng.shuffle(matched)
    report = {
        "requested_target_per_class": target_per_class,
        "actual_per_class": actual_per_class,
        "bin_width": bin_width,
        "random_seed": random_seed,
        "available_by_task": dict(Counter(ex["task_type"] for ex in examples)),
        "selected_by_task": {task: len(vals) for task, vals in selected_by_task.items()},
        "discarded_by_task": {
            task: Counter(ex["task_type"] for ex in examples)[task] - len(selected_by_task[task])
            for task in [REASONING_TASK, FACTUAL_RECALL_TASK]
        },
        "capacity_by_bin": capacity_by_bin,
        "selected_per_bin": selected_per_bin,
    }
    print("Distributional length matching report")
    print(json.dumps(report, indent=2, default=expanded_json_default))
    if actual_per_class < target_per_class:
        print("WARNING: requested target_per_class was not possible under overlapping length bins.")
    return matched, report


def expanded_verify_length_match(original_examples, matched_examples, plot=True, save_dir=None):
    original_stats = expanded_analyze_lengths(
        original_examples,
        title="Before expanded length matching",
        plot=plot,
        save_dir=save_dir,
    )
    matched_stats = expanded_analyze_lengths(
        matched_examples,
        title="After expanded length matching",
        plot=plot,
        save_dir=save_dir,
    )
    matched_df = matched_stats["table"]
    r = matched_df.loc[matched_df["task_type"] == REASONING_TASK, "token_length"].to_numpy()
    f = matched_df.loc[matched_df["task_type"] == FACTUAL_RECALL_TASK, "token_length"].to_numpy()
    smd = abs(np.mean(r) - np.mean(f)) / np.sqrt((np.var(r, ddof=1) + np.var(f, ddof=1)) / 2)
    diff = {
        "absolute_mean_difference": float(abs(np.mean(r) - np.mean(f))),
        "absolute_median_difference": float(abs(np.median(r) - np.median(f))),
        "standardized_mean_difference": float(smd),
    }
    print("Length-match difference summary")
    print(json.dumps(diff, indent=2))
    if diff["standardized_mean_difference"] > 0.25:
        print("WARNING: standardized mean difference remains above 0.25.")
    return {
        "before": original_stats,
        "after": matched_stats,
        "differences": diff,
    }
````

````python
# Cross-validation metrics and baseline/probe helpers

def expanded_metric_summary(values):
    values = np.asarray(values, dtype=float)
    return {
        "mean": float(values.mean()),
        "std": float(values.std()),
        "values": values.tolist(),
    }


def expanded_binary_cv(X, y, title, random_state=42, n_splits=5):
    # Keep scipy sparse matrices sparse. Calling np.asarray() on a
    # csr_matrix creates a 0-dimensional object array, which breaks
    # StratifiedKFold and sklearn indexing for the TF-IDF baseline.
    if hasattr(X, "tocsr"):
        X = X.tocsr()
    else:
        X = np.asarray(X)

    y = np.asarray(y, dtype=int)
    x_len = X.shape[0] if hasattr(X, "shape") else len(X)
    if x_len != len(y):
        raise ValueError(f"X/y length mismatch: X has {x_len} rows, y has {len(y)} labels.")

    min_count = min(Counter(y).values())
    if min_count < n_splits:
        raise ValueError(f"Need at least {n_splits} examples per class, got min class count {min_count}.")

    cv = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=random_state)
    rows = []
    print("\n" + title)
    print("=" * len(title))
    for fold, (train_idx, test_idx) in enumerate(cv.split(X, y), start=1):
        clf = LogisticRegression(C=1.0, max_iter=5000)
        clf.fit(X[train_idx], y[train_idx])
        pred = clf.predict(X[test_idx])
        prob = clf.predict_proba(X[test_idx])[:, 1]
        row = {
            "fold": fold,
            "accuracy": float(accuracy_score(y[test_idx], pred)),
            "balanced_accuracy": float(balanced_accuracy_score(y[test_idx], pred)),
            "auroc": float(roc_auc_score(y[test_idx], prob)),
        }
        rows.append(row)
        print(f"Fold {fold}: acc={row['accuracy']:.4f}, bal_acc={row['balanced_accuracy']:.4f}, auroc={row['auroc']:.4f}")
    result = {
        "title": title,
        "folds": rows,
        "accuracy": expanded_metric_summary([r["accuracy"] for r in rows]),
        "balanced_accuracy": expanded_metric_summary([r["balanced_accuracy"] for r in rows]),
        "auroc": expanded_metric_summary([r["auroc"] for r in rows]),
    }
    print("Summary:")
    print(f"  accuracy          {result['accuracy']['mean']:.4f} +/- {result['accuracy']['std']:.4f}")
    print(f"  balanced_accuracy {result['balanced_accuracy']['mean']:.4f} +/- {result['balanced_accuracy']['std']:.4f}")
    print(f"  auroc             {result['auroc']['mean']:.4f} +/- {result['auroc']['std']:.4f}")
    return result


def expanded_length_only_baseline(matched_examples):
    X = np.asarray([ex["token_length"] for ex in matched_examples], dtype=np.float32).reshape(-1, 1)
    y = np.asarray([TASK_TO_LABEL[ex["task_type"]] for ex in matched_examples], dtype=int)
    result = expanded_binary_cv(X, y, "Expanded length-only baseline")
    if result["accuracy"]["mean"] > 0.75 or result["auroc"]["mean"] > 0.80:
        print("WARNING: token length remains predictive after matching.")
    return result


def expanded_tfidf_baseline(matched_examples, max_features=20000, ngram_range=(1, 2), random_state=42):
    texts = [ex["question"] for ex in matched_examples]
    y = np.asarray([TASK_TO_LABEL[ex["task_type"]] for ex in matched_examples], dtype=int)
    vectorizer = TfidfVectorizer(
        lowercase=True,
        strip_accents="unicode",
        max_features=max_features,
        ngram_range=ngram_range,
        min_df=2,
    )
    X = vectorizer.fit_transform(texts)
    result = expanded_binary_cv(X, y, "Expanded TF-IDF lexical baseline", random_state=random_state)
    result["tfidf_config"] = {
        "max_features": max_features,
        "ngram_range": list(ngram_range),
        "min_df": 2,
        "vocabulary_size": len(vectorizer.vocabulary_),
    }
    if result["accuracy"]["mean"] > 0.75 or result["auroc"]["mean"] > 0.80:
        print("WARNING: lexical features remain strongly predictive.")
    return result, vectorizer


def expanded_random_label_probe(X, y, repeats=10, random_seed=123):
    rng = np.random.default_rng(random_seed)
    repeat_results = []
    for i in range(repeats):
        shuffled = np.asarray(y, dtype=int).copy()
        rng.shuffle(shuffled)
        repeat_results.append(expanded_binary_cv(X, shuffled, f"Random-label probe repeat {i + 1}", random_state=42))
    result = {
        "repeats": repeats,
        "random_seed": random_seed,
        "accuracy": expanded_metric_summary([r["accuracy"]["mean"] for r in repeat_results]),
        "balanced_accuracy": expanded_metric_summary([r["balanced_accuracy"]["mean"] for r in repeat_results]),
        "auroc": expanded_metric_summary([r["auroc"]["mean"] for r in repeat_results]),
        "repeat_results": repeat_results,
    }
    print("Random-label repeated summary")
    print(json.dumps({
        "accuracy": result["accuracy"],
        "balanced_accuracy": result["balanced_accuracy"],
        "auroc": result["auroc"],
    }, indent=2, default=expanded_json_default))
    return result
````

````python
# Qwen hidden-state extraction and representation loading/saving

def expanded_extract_qwen_hidden_states(matched_examples, model, tokenizer, paths, device=None, force_recompute=False):
    hidden_path = paths["representations"] / "hidden_states.pt"
    final_path = paths["representations"] / "final_layer.pt"
    metadata_path = paths["representations"] / "metadata.json"
    if hidden_path.exists() and final_path.exists() and metadata_path.exists() and not force_recompute:
        print("Using cached expanded representations:")
        print(" ", hidden_path)
        hidden_payload = torch.load(hidden_path, map_location="cpu")
        final_payload = torch.load(final_path, map_location="cpu")
        metadata = expanded_load_json(metadata_path)
        return {
            "hidden_states": hidden_payload["hidden_states"],
            "final_layer": final_payload["final_layer"],
            "metadata": metadata,
            "config": hidden_payload.get("config", {}),
            "paths": {
                "hidden_states": str(hidden_path),
                "final_layer": str(final_path),
                "metadata": str(metadata_path),
            },
        }
    if device is None:
        device = "cuda" if torch.cuda.is_available() else "cpu"
    model.to(device)
    model.eval()
    all_hidden = []
    metadata = []
    with torch.no_grad():
        for idx, ex in enumerate(matched_examples):
            prompt = ex["formatted_prompt"]
            input_ids = torch.tensor(tokenizer.encode(prompt), dtype=torch.long, device=device).unsqueeze(0)
            _, hidden_layers = model(input_ids, cache=None)
            hidden_layers = hidden_layers.squeeze(0).detach().cpu()
            all_hidden.append(hidden_layers)
            row = deepcopy(ex)
            row["tensor_row"] = idx
            metadata.append(row)
            if (idx + 1) % 25 == 0:
                print(f"Extracted {idx + 1}/{len(matched_examples)} examples")
    hidden_states = torch.stack(all_hidden, dim=0)
    final_layer = hidden_states[:, -1, :]
    config = {
        "model": "Qwen3-0.6B custom implementation",
        "tokenizer": "custom Qwen3 tokenizer",
        "prompt_template_version": EXPANDED_PROMPT_TEMPLATE_VERSION,
        "representation": "final_formatted_prompt_token_after_each_transformer_block",
        "primary_layer": "final",
        "use_kv_cache": False,
        "generation": False,
        "shape_hidden_states": list(hidden_states.shape),
        "shape_final_layer": list(final_layer.shape),
    }
    torch.save({"hidden_states": hidden_states, "config": config}, hidden_path)
    torch.save({"final_layer": final_layer, "config": config}, final_path)
    expanded_save_json(metadata, metadata_path)
    print("Saved representations:")
    print(" ", hidden_path, tuple(hidden_states.shape))
    print(" ", final_path, tuple(final_layer.shape))
    print(" ", metadata_path)
    return {
        "hidden_states": hidden_states,
        "final_layer": final_layer,
        "metadata": metadata,
        "config": config,
        "paths": {
            "hidden_states": str(hidden_path),
            "final_layer": str(final_path),
            "metadata": str(metadata_path),
        },
    }


def expanded_representation_xy(representation_payload):
    X = representation_payload["final_layer"].float().cpu().numpy()
    y = np.asarray([TASK_TO_LABEL[ex["task_type"]] for ex in representation_payload["metadata"]], dtype=int)
    return X, y
````

````python
# PCA, main probe, cross-dataset evaluation, per-dataset analysis, and multi-seed robustness

def expanded_run_pca(X, y, metadata, paths):
    pca = PCA(n_components=2)
    X_pca = pca.fit_transform(X)
    explained = pca.explained_variance_ratio_
    result = {
        "pc1_variance": float(explained[0]),
        "pc2_variance": float(explained[1]),
        "pc1_pc2_variance": float(explained.sum()),
    }
    print("Expanded final-layer PCA")
    print(json.dumps(result, indent=2))
    fig, ax = plt.subplots(figsize=(8, 6))
    for task, label in TASK_TO_LABEL.items():
        mask = y == label
        ax.scatter(X_pca[mask, 0], X_pca[mask, 1], alpha=0.7, label=task)
    ax.set_xlabel("PC1")
    ax.set_ylabel("PC2")
    ax.set_title("Expanded length-controlled final-layer PCA")
    ax.legend()
    plt.tight_layout()
    fig_path = paths["fig_pca"] / "expanded_final_layer_pca.png"
    fig.savefig(fig_path, dpi=200)
    plt.show()
    np.savez(paths["pca"] / "expanded_final_layer_pca.npz", X_pca=X_pca, y=y, **result)
    result["figure_path"] = str(fig_path)
    result["npz_path"] = str(paths["pca"] / "expanded_final_layer_pca.npz")
    return result


def expanded_qwen_final_layer_probe(X, y):
    return expanded_binary_cv(X, y, "Expanded Qwen final-layer probe", random_state=42)


def expanded_cross_dataset_eval(X, y, metadata):
    dataset = np.asarray([ex["dataset"] for ex in metadata])
    directions = [
        {
            "name": "train_math500_popqa_test_gsm8k_triviaqa",
            "train_datasets": {"math500", "popqa"},
            "test_datasets": {"gsm8k", "triviaqa"},
        },
        {
            "name": "train_gsm8k_triviaqa_test_math500_popqa",
            "train_datasets": {"gsm8k", "triviaqa"},
            "test_datasets": {"math500", "popqa"},
        },
    ]
    results = []
    for spec in directions:
        train_mask = np.isin(dataset, list(spec["train_datasets"]))
        test_mask = np.isin(dataset, list(spec["test_datasets"]))
        y_train = y[train_mask]
        y_test = y[test_mask]
        if len(set(y_train.tolist())) < 2 or len(set(y_test.tolist())) < 2:
            print("Skipping cross-dataset direction due to missing class:", spec["name"])
            continue
        clf = LogisticRegression(C=1.0, max_iter=5000)
        clf.fit(X[train_mask], y_train)
        pred = clf.predict(X[test_mask])
        prob = clf.predict_proba(X[test_mask])[:, 1]
        result = {
            "name": spec["name"],
            "train_datasets": sorted(spec["train_datasets"]),
            "test_datasets": sorted(spec["test_datasets"]),
            "train_count": int(train_mask.sum()),
            "test_count": int(test_mask.sum()),
            "train_class_counts": dict(Counter(y_train.tolist())),
            "test_class_counts": dict(Counter(y_test.tolist())),
            "accuracy": float(accuracy_score(y_test, pred)),
            "balanced_accuracy": float(balanced_accuracy_score(y_test, pred)),
            "auroc": float(roc_auc_score(y_test, prob)),
        }
        print("Cross-dataset result:", json.dumps(result, indent=2, default=expanded_json_default))
        results.append(result)
    return results


def expanded_per_dataset_pair_analysis(X, y, metadata):
    dataset = np.asarray([ex["dataset"] for ex in metadata])
    pairs = [
        ("math500", "popqa"),
        ("math500", "triviaqa"),
        ("gsm8k", "popqa"),
        ("gsm8k", "triviaqa"),
    ]
    results = {}
    for reasoning_ds, factual_ds in pairs:
        mask = np.isin(dataset, [reasoning_ds, factual_ds])
        if mask.sum() == 0:
            continue
        y_pair = y[mask]
        counts = Counter(y_pair.tolist())
        if len(counts) < 2 or min(counts.values()) < 5:
            print(f"Skipping {reasoning_ds} vs {factual_ds}; insufficient class counts:", counts)
            continue
        title = f"Per-dataset probe: {reasoning_ds} vs {factual_ds}"
        results[f"{reasoning_ds}_vs_{factual_ds}"] = expanded_binary_cv(X[mask], y_pair, title)
    return results


def expanded_multi_seed_probe(X, y, seeds=(42, 123, 456)):
    seed_results = {}
    for seed in seeds:
        seed_results[str(seed)] = expanded_binary_cv(X, y, f"Expanded Qwen final-layer probe seed {seed}", random_state=seed)
    summary = {
        "seeds": list(seeds),
        "accuracy": expanded_metric_summary([r["accuracy"]["mean"] for r in seed_results.values()]),
        "balanced_accuracy": expanded_metric_summary([r["balanced_accuracy"]["mean"] for r in seed_results.values()]),
        "auroc": expanded_metric_summary([r["auroc"]["mean"] for r in seed_results.values()]),
        "seed_results": seed_results,
    }
    print("Multi-seed probe summary")
    print(json.dumps({
        "accuracy": summary["accuracy"],
        "balanced_accuracy": summary["balanced_accuracy"],
        "auroc": summary["auroc"],
    }, indent=2, default=expanded_json_default))
    return summary
````

````python
# Summary table, artifact saving, and final report

def expanded_make_baseline_summary(length_result, tfidf_result, qwen_result, random_label_result):
    rows = [
        {
            "method": "Length-only",
            "accuracy": f"{length_result['accuracy']['mean']:.4f} +/- {length_result['accuracy']['std']:.4f}",
            "auroc": f"{length_result['auroc']['mean']:.4f} +/- {length_result['auroc']['std']:.4f}",
        },
        {
            "method": "TF-IDF lexical",
            "accuracy": f"{tfidf_result['accuracy']['mean']:.4f} +/- {tfidf_result['accuracy']['std']:.4f}",
            "auroc": f"{tfidf_result['auroc']['mean']:.4f} +/- {tfidf_result['auroc']['std']:.4f}",
        },
        {
            "method": "Qwen final-layer probe",
            "accuracy": f"{qwen_result['accuracy']['mean']:.4f} +/- {qwen_result['accuracy']['std']:.4f}",
            "auroc": f"{qwen_result['auroc']['mean']:.4f} +/- {qwen_result['auroc']['std']:.4f}",
        },
        {
            "method": "Random-label probe",
            "accuracy": f"{random_label_result['accuracy']['mean']:.4f} +/- {random_label_result['accuracy']['std']:.4f}",
            "auroc": f"{random_label_result['auroc']['mean']:.4f} +/- {random_label_result['auroc']['std']:.4f}",
        },
    ]
    df = pd.DataFrame(rows)
    display(df)
    return df


def expanded_save_all_results(paths, result_bundle):
    saved = {}
    for name, obj in result_bundle.items():
        if isinstance(obj, pd.DataFrame):
            path = paths["baselines"] / f"{name}.csv"
            obj.to_csv(path, index=False)
        else:
            if name in {"pca_results"}:
                path = paths["pca"] / f"{name}.json"
            elif name in {"cross_dataset_results"}:
                path = paths["cross_dataset"] / f"{name}.json"
            elif name in {"per_dataset_results"}:
                path = paths["per_dataset"] / f"{name}.json"
            elif name in {"qwen_probe_results", "random_label_results", "multi_seed_results"}:
                path = paths["linear_probe"] / f"{name}.json"
            else:
                path = paths["baselines"] / f"{name}.json"
            expanded_save_json(obj, path)
        saved[name] = str(path)
    print("Saved result files")
    for key, path in saved.items():
        print(f"  {key}: {path}")
    return saved


def expanded_print_final_report(
    raw_examples,
    deduped_examples,
    removed_duplicates,
    matched_examples,
    match_report,
    length_verification,
    length_result,
    tfidf_result,
    qwen_result,
    random_label_result,
    cross_dataset_results,
    per_dataset_results,
    multi_seed_results,
    pca_results,
    saved_paths,
    representation_paths,
):
    print("\n" + "=" * 70)
    print("EXPANDED RQ1 CONTROL EXPERIMENT REPORT")
    print("=" * 70)
    print("1. Dataset counts before/after deduplication")
    print("   raw examples      :", len(raw_examples))
    print("   deduped examples  :", len(deduped_examples))
    print("   removed duplicates:", len(removed_duplicates))
    print("2. Counts per source dataset")
    print("   raw    :", dict(Counter(ex["dataset"] for ex in raw_examples)))
    print("   deduped:", dict(Counter(ex["dataset"] for ex in deduped_examples)))
    print("   matched:", dict(Counter(ex["dataset"] for ex in matched_examples)))
    print("3. Final matched dataset size")
    print("   matched total:", len(matched_examples))
    print("   matched by task:", dict(Counter(ex["task_type"] for ex in matched_examples)))
    print("   match report actual_per_class:", match_report["actual_per_class"])
    print("4. Token-length statistics")
    print("   differences:", length_verification["differences"])
    print("5. Length-only baseline")
    print("   accuracy:", length_result["accuracy"])
    print("   auroc   :", length_result["auroc"])
    print("6. TF-IDF lexical baseline")
    print("   accuracy:", tfidf_result["accuracy"])
    print("   auroc   :", tfidf_result["auroc"])
    print("7. Qwen final-layer probe")
    print("   accuracy:", qwen_result["accuracy"])
    print("   auroc   :", qwen_result["auroc"])
    print("8. Random-label control")
    print("   accuracy:", random_label_result["accuracy"])
    print("   auroc   :", random_label_result["auroc"])
    print("9. Cross-dataset results")
    print(json.dumps(cross_dataset_results, indent=2, default=expanded_json_default))
    print("10. Per-dataset results")
    print(json.dumps(per_dataset_results, indent=2, default=expanded_json_default))
    print("11. PCA explained variance")
    print(json.dumps(pca_results, indent=2, default=expanded_json_default))
    print("12. Saved artifacts")
    print("   representation paths:", representation_paths)
    print("   result paths:", saved_paths)
    print("\nSCIENTIFIC INTERPRETATION")
    print("- Representations may distinguish the task groups if Qwen probe performance exceeds length-only and lexical baselines.")
    print("- Strong representation separation does not by itself prove the model contains a reasoning representation.")
    print("- This experiment does not test or claim that reasoning is encoded in one linear direction.")
    print("- Proceed to the explicit single-direction test only after inspecting these controls.")
````
