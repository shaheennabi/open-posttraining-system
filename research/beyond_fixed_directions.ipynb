### Beyond Fixed Directions: Understanding the Representational Geometry of Reasoning and Memorization in Large Language Models

---

#### Core Scientific Question

**How are reasoning and memorization represented inside large language models, and how do these representations evolve during reinforcement learning?**

Rather than assuming reasoning and memorization can be represented by a single fixed direction, this work systematically studies:

- Their internal geometry
- Their evolution during reinforcement learning
- Their ability to generalize across tasks
- Their internal mechanistic composition

> **DiRL is an applied motivation, not the subject of the paper.**

---

#### Research Hypothesis

Reasoning and memorization may **not** be adequately represented by a single fixed linear direction.

Instead, they may emerge as **adaptive, evolving, multi-component latent representations** whose geometry changes during reinforcement learning and varies across tasks.

This is the hypothesis that the experiments are designed to test—not a conclusion assumed beforehand.

---

#### Research Objectives

1. Characterize the geometry of reasoning and memorization.
2. Study how these representations evolve during reinforcement learning.
3. Understand whether they generalize across different reasoning tasks.
4. Identify the latent components that compose reasoning and memorization.

````python
### the tokenizer also adds the formatted assistant/special tokens
### before generation
###
### so for our primary experiment we take the representation
### of the LAST FORMATTED PROMPT TOKEN


def get100(problems, device, prompt_key="problem"):  ## prompt_key as --> if you have memorization data pass question, otherwise problem for reasoning dataset

    ## store hidden representations for all 100 problems
    all_hidden_layers = []

    model.eval()  ## get the model in the eval mode.. it will add ..--->  disabling layers like dropout, and using running batchnorm statistics from training

    with torch.no_grad():

        ## go problem-by-problem
        for i in problems:
            
            ## get current problem
            prompt = i[prompt_key]  ## this should work for both the reasoning and memoriation data


            ## tokenize the complete formatted prompt
            ## then convert to tensor
            ##
            ## shape --> (1, seq_len)
            input_ids = torch.tensor(tokenizer.encode(prompt), dtype=torch.long, device=device).unsqueeze(0)


            ## forward pass
            ##
            ## hidden_layers shape:
            ## (1, n_layers, emb_dim)
            _ , hidden_layers = model(input_ids, cache=None)


            ## remove batch dimension
            ##
            ## (1, n_layers, emb_dim)
            ##          ↓
            ## (n_layers, emb_dim)
            hidden_layers = hidden_layers.squeeze(0)


            ## move representations to CPU
            ## and store current problem
            all_hidden_layers.append(hidden_layers.detach().cpu())


    ## after processing all 100 problems:
    ##
    ## list:
    ##
    ## problem 1   --> (n_layers, emb_dim)
    ## problem 2   --> (n_layers, emb_dim)
    ## ...
    ## problem 100 --> (n_layers, emb_dim)
    ##
    ## stack them:
    ##
    ## (100, n_layers, emb_dim)
    all_hidden_layers = torch.stack(all_hidden_layers, dim=0)


    return all_hidden_layers
````

## Length-matched MATH-500 vs PopQA control experiment

This section controls the strong prompt-token-length confound found in the original MATH-500 vs PopQA comparison. It does not modify question content. Each helper is defined directly in this notebook so the experiment is self-contained for Lightning AI Studio.

Run the first runner to build/verify the matched subset. Run the GPU runner only after the length-only baseline is acceptable. The extraction uses the existing custom `model` and `tokenizer`; it does not replace them with Hugging Face implementations.


````python
# Length-matched control experiment imports and constants
from copy import deepcopy
from pathlib import Path
import json
import math

import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt

from sklearn.decomposition import PCA
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, balanced_accuracy_score, roc_auc_score
from sklearn.model_selection import StratifiedKFold


REASONING_LABEL = 1
MEMORIZATION_LABEL = 0
````

````python
def get_prompt_text(example, prompt_key=None):
    """Return the text field used by the existing extraction pipeline."""
    if prompt_key is not None:
        return example[prompt_key]
    if "problem" in example:
        return example["problem"]
    if "question" in example:
        return example["question"]
    raise KeyError("Example must contain either 'problem' or 'question', or pass prompt_key.")
````

````python
def token_length_after_pipeline(example, tokenizer, prompt_key=None):
    """Measure length after the same tokenizer.encode(prompt) call used for hidden extraction."""
    prompt = get_prompt_text(example, prompt_key=prompt_key)
    return len(tokenizer.encode(prompt))
````

````python
def _summary_stats(values):
    values = np.asarray(values, dtype=float)
    return {
        "count": int(values.size),
        "min": float(np.min(values)),
        "max": float(np.max(values)),
        "mean": float(np.mean(values)),
        "median": float(np.median(values)),
        "std": float(np.std(values, ddof=1)) if values.size > 1 else 0.0,
        "p25": float(np.percentile(values, 25)),
        "p50": float(np.percentile(values, 50)),
        "p75": float(np.percentile(values, 75)),
    }
````

````python
def _print_stats(name, values):
    stats = _summary_stats(values)
    print(f"\n{name}")
    print("-" * len(name))
    for key in ["count", "min", "max", "mean", "median", "std", "p25", "p50", "p75"]:
        value = stats[key]
        if key == "count":
            print(f"{key:>8}: {value}")
        else:
            print(f"{key:>8}: {value:.3f}")
    return stats
````

````python
def analyze_token_lengths(
    reasoning_examples,
    memorization_examples,
    tokenizer,
    reasoning_prompt_key="problem",
    memorization_prompt_key="question",
    plot=True,
):
    """Tokenize both original datasets and summarize their prompt-token lengths."""
    records = []

    for index, example in enumerate(reasoning_examples):
        token_length = token_length_after_pipeline(example, tokenizer, reasoning_prompt_key)
        records.append(
            {
                "index": index,
                "id": example.get("id", index),
                "dataset": example.get("dataset", "MATH-500"),
                "label": "reasoning",
                "label_id": REASONING_LABEL,
                "token_length": token_length,
            }
        )

    for index, example in enumerate(memorization_examples):
        token_length = token_length_after_pipeline(example, tokenizer, memorization_prompt_key)
        records.append(
            {
                "index": index,
                "id": example.get("id", index),
                "dataset": example.get("dataset", "PopQA"),
                "label": "memorization",
                "label_id": MEMORIZATION_LABEL,
                "token_length": token_length,
            }
        )

    length_df = pd.DataFrame(records)
    reasoning_lengths = length_df.loc[length_df["label"] == "reasoning", "token_length"].to_numpy()
    memorization_lengths = length_df.loc[length_df["label"] == "memorization", "token_length"].to_numpy()

    print("Original token-length distributions")
    print("=" * 44)
    reasoning_stats = _print_stats("Reasoning", reasoning_lengths)
    memorization_stats = _print_stats("Memorization", memorization_lengths)

    if plot:
        plt.figure(figsize=(9, 5))
        plt.hist(reasoning_lengths, bins=20, alpha=0.60, label="Reasoning")
        plt.hist(memorization_lengths, bins=20, alpha=0.60, label="Memorization")
        plt.xlabel("Prompt token length")
        plt.ylabel("Example count")
        plt.title("Before length matching")
        plt.legend()
        plt.tight_layout()
        plt.show()

    return {
        "length_df": length_df,
        "reasoning_stats": reasoning_stats,
        "memorization_stats": memorization_stats,
    }
````

````python
def _maximum_bipartite_matching(reasoning_lengths, memorization_lengths, max_length_difference):
    """Maximum-cardinality one-to-one matching within the length tolerance."""
    edges = []
    for r_idx, r_len in enumerate(reasoning_lengths):
        possible = []
        for m_idx, m_len in enumerate(memorization_lengths):
            diff = abs(int(r_len) - int(m_len))
            if diff <= max_length_difference:
                possible.append((diff, m_idx))
        edges.append([m_idx for _, m_idx in sorted(possible)])

    match_mem_to_reason = {}

    def try_assign(r_idx, seen):
        for m_idx in edges[r_idx]:
            if m_idx in seen:
                continue
            seen.add(m_idx)
            if m_idx not in match_mem_to_reason or try_assign(match_mem_to_reason[m_idx], seen):
                match_mem_to_reason[m_idx] = r_idx
                return True
        return False

    for r_idx in np.argsort(reasoning_lengths):
        try_assign(int(r_idx), set())

    pairs = [(r_idx, m_idx) for m_idx, r_idx in match_mem_to_reason.items()]
    pairs.sort(key=lambda pair: (reasoning_lengths[pair[0]], pair[0]))
    return pairs
````

````python
def match_by_token_length(
    reasoning_examples,
    memorization_examples,
    tokenizer,
    max_length_difference=5,
    reasoning_prompt_key="problem",
    memorization_prompt_key="question",
):
    """Create a new one-to-one length-matched dataset without modifying the originals."""
    reasoning_lengths = np.array(
        [token_length_after_pipeline(example, tokenizer, reasoning_prompt_key) for example in reasoning_examples]
    )
    memorization_lengths = np.array(
        [token_length_after_pipeline(example, tokenizer, memorization_prompt_key) for example in memorization_examples]
    )

    pairs = _maximum_bipartite_matching(reasoning_lengths, memorization_lengths, max_length_difference)
    matched_reasoning = []
    matched_memorization = []

    for match_id, (r_idx, m_idx) in enumerate(pairs, start=1):
        r_example = deepcopy(reasoning_examples[r_idx])
        m_example = deepcopy(memorization_examples[m_idx])
        r_len = int(reasoning_lengths[r_idx])
        m_len = int(memorization_lengths[m_idx])
        diff = abs(r_len - m_len)

        r_example.update(
            {
                "dataset": r_example.get("dataset", "MATH-500"),
                "label": "reasoning",
                "label_id": REASONING_LABEL,
                "token_length": r_len,
                "match_id": match_id,
                "matched_with_label": "memorization",
                "matched_with_original_index": int(m_idx),
                "matched_with_id": memorization_examples[m_idx].get("id", int(m_idx)),
                "length_difference": int(diff),
                "max_length_difference": int(max_length_difference),
            }
        )
        m_example.update(
            {
                "dataset": m_example.get("dataset", "PopQA"),
                "label": "memorization",
                "label_id": MEMORIZATION_LABEL,
                "token_length": m_len,
                "match_id": match_id,
                "matched_with_label": "reasoning",
                "matched_with_original_index": int(r_idx),
                "matched_with_id": reasoning_examples[r_idx].get("id", int(r_idx)),
                "length_difference": int(diff),
                "max_length_difference": int(max_length_difference),
            }
        )
        matched_reasoning.append(r_example)
        matched_memorization.append(m_example)

    print("\nLength matching result")
    print("=" * 44)
    print("max_length_difference       :", max_length_difference)
    print("original reasoning count    :", len(reasoning_examples))
    print("original memorization count :", len(memorization_examples))
    print("matched reasoning count     :", len(matched_reasoning))
    print("matched memorization count  :", len(matched_memorization))
    print("discarded reasoning count   :", len(reasoning_examples) - len(matched_reasoning))
    print("discarded memorization count:", len(memorization_examples) - len(matched_memorization))

    if len(matched_reasoning) < min(len(reasoning_examples), len(memorization_examples)):
        print("WARNING: full 100-vs-100 matching was not possible at this tolerance.")

    return {
        "matched_reasoning": matched_reasoning,
        "matched_memorization": matched_memorization,
        "matched_dataset": matched_reasoning + matched_memorization,
        "pairs": pairs,
        "reasoning_lengths": reasoning_lengths,
        "memorization_lengths": memorization_lengths,
    }
````

````python
def _as_metric_summary(values):
    values = np.asarray(values, dtype=float)
    return {"mean": float(values.mean()), "std": float(values.std()), "values": values.tolist()}
````

````python
def _ensure_5_fold_possible(y):
    counts = np.bincount(np.asarray(y, dtype=int))
    min_class_count = int(counts[counts > 0].min())
    if min_class_count < 5:
        raise ValueError(f"Need at least 5 examples per class for 5-fold CV, got {min_class_count}.")
````

````python
def run_linear_probe(X, y, title="Linear probe", random_state=42):
    """Run the same 5-fold LogisticRegression probe used in the original notebook."""
    X = np.asarray(X, dtype=np.float32)
    y = np.asarray(y, dtype=int)
    _ensure_5_fold_possible(y)

    cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=random_state)
    accuracies = []
    balanced_accuracies = []
    aurocs = []
    fold_rows = []

    print("\n" + title)
    print("=" * len(title))
    for fold, (train_idx, test_idx) in enumerate(cv.split(X, y), start=1):
        probe = LogisticRegression(C=1.0, max_iter=5000)
        probe.fit(X[train_idx], y[train_idx])
        y_pred = probe.predict(X[test_idx])
        y_prob = probe.predict_proba(X[test_idx])[:, 1]

        accuracy = accuracy_score(y[test_idx], y_pred)
        balanced_accuracy = balanced_accuracy_score(y[test_idx], y_pred)
        auroc = roc_auc_score(y[test_idx], y_prob)

        accuracies.append(accuracy)
        balanced_accuracies.append(balanced_accuracy)
        aurocs.append(auroc)
        fold_rows.append(
            {
                "fold": fold,
                "accuracy": float(accuracy),
                "balanced_accuracy": float(balanced_accuracy),
                "auroc": float(auroc),
            }
        )
        print(f"Fold {fold}: accuracy={accuracy:.4f}, balanced_accuracy={balanced_accuracy:.4f}, auroc={auroc:.4f}")

    results = {
        "folds": fold_rows,
        "accuracy": _as_metric_summary(accuracies),
        "balanced_accuracy": _as_metric_summary(balanced_accuracies),
        "auroc": _as_metric_summary(aurocs),
    }
    print("\nSummary")
    print("-" * 20)
    print(f"Accuracy          : {results['accuracy']['mean']:.4f} +/- {results['accuracy']['std']:.4f}")
    print(f"Balanced Accuracy : {results['balanced_accuracy']['mean']:.4f} +/- {results['balanced_accuracy']['std']:.4f}")
    print(f"AUROC             : {results['auroc']['mean']:.4f} +/- {results['auroc']['std']:.4f}")
    return results
````

````python
def length_only_baseline(matched_reasoning, matched_memorization, warning_accuracy=0.80, warning_auroc=0.80):
    """Classify reasoning vs memorization using only prompt token length."""
    token_lengths = [x["token_length"] for x in matched_reasoning] + [x["token_length"] for x in matched_memorization]
    X_length = np.asarray(token_lengths, dtype=np.float32).reshape(-1, 1)
    y = np.asarray([REASONING_LABEL] * len(matched_reasoning) + [MEMORIZATION_LABEL] * len(matched_memorization), dtype=int)
    results = run_linear_probe(X_length, y, title="Length-only baseline")
    if results["accuracy"]["mean"] >= warning_accuracy or results["auroc"]["mean"] >= warning_auroc:
        print("\nWARNING: token length is still strongly predictive on the matched dataset.")
        print("Do not interpret hidden-state probe results until this confound is reduced.")
    return results
````

````python
def run_pca(X, y, title="Final-layer PCA", plot=True):
    """Fit PCA without using labels, then use labels only for plotting."""
    X = np.asarray(X, dtype=np.float32)
    y = np.asarray(y, dtype=int)
    pca = PCA(n_components=2)
    X_pca = pca.fit_transform(X)
    explained = pca.explained_variance_ratio_
    print("\n" + title)
    print("=" * len(title))
    print(f"PC1 explained variance ratio     : {explained[0]:.4f}")
    print(f"PC2 explained variance ratio     : {explained[1]:.4f}")
    print(f"PC1 + PC2 explained variance     : {explained.sum():.4f}")
    if plot:
        plt.figure(figsize=(8, 6))
        plt.scatter(X_pca[y == REASONING_LABEL, 0], X_pca[y == REASONING_LABEL, 1], label="Reasoning", alpha=0.75)
        plt.scatter(
            X_pca[y == MEMORIZATION_LABEL, 0],
            X_pca[y == MEMORIZATION_LABEL, 1],
            label="Memorization",
            alpha=0.75,
        )
        plt.xlabel("PC1")
        plt.ylabel("PC2")
        plt.title(title)
        plt.legend()
        plt.tight_layout()
        plt.show()
    return {
        "pca": pca,
        "X_pca": X_pca,
        "pc1_variance": float(explained[0]),
        "pc2_variance": float(explained[1]),
        "pc1_pc2_variance": float(explained.sum()),
    }
````

````python
def run_random_label_baseline(X, y, random_state=123):
    """Shuffle labels while keeping X unchanged, then rerun the same 5-fold probe."""
    rng = np.random.default_rng(random_state)
    shuffled_y = np.asarray(y, dtype=int).copy()
    rng.shuffle(shuffled_y)
    return run_linear_probe(X, shuffled_y, title="Random-label control", random_state=42)
````

````python
def compare_original_vs_matched(length_results, hidden_probe_results, pca_results):
    comparison = pd.DataFrame(
        [
            {"metric": "Length-only accuracy", "Original": 0.965, "Length-matched": length_results["accuracy"]["mean"]},
            {"metric": "Length-only AUROC", "Original": 0.994, "Length-matched": length_results["auroc"]["mean"]},
            {"metric": "Hidden probe accuracy", "Original": 1.000, "Length-matched": hidden_probe_results["accuracy"]["mean"]},
            {"metric": "Hidden probe AUROC", "Original": 1.000, "Length-matched": hidden_probe_results["auroc"]["mean"]},
            {"metric": "PCA PC1 variance", "Original": 0.4904, "Length-matched": pca_results["pc1_variance"]},
            {"metric": "PCA PC2 variance", "Original": 0.0888, "Length-matched": pca_results["pc2_variance"]},
        ]
    )
    print("\nOriginal vs length-matched comparison")
    print("=" * 44)
    try:
        display(comparison)
    except NameError:
        print(comparison.to_string(index=False))
    return comparison
````

````python
def extract_matched_hidden_states(
    matched_reasoning,
    matched_memorization,
    model,
    tokenizer,
    device=None,
    reasoning_prompt_key="problem",
    memorization_prompt_key="question",
):
    """Extract [N, 28, 1024] hidden states for the matched dataset using the existing Qwen3 model."""
    if device is None:
        device = "cuda" if torch.cuda.is_available() else "cpu"
    model.to(device)
    model.eval()

    def extract_group(examples, prompt_key):
        all_hidden_layers = []
        with torch.no_grad():
            for example in examples:
                prompt = get_prompt_text(example, prompt_key=prompt_key)
                input_ids = torch.tensor(tokenizer.encode(prompt), dtype=torch.long, device=device).unsqueeze(0)
                _, hidden_layers = model(input_ids, cache=None)
                all_hidden_layers.append(hidden_layers.squeeze(0).detach().cpu())
        return torch.stack(all_hidden_layers, dim=0)

    reasoning_hidden = extract_group(matched_reasoning, reasoning_prompt_key)
    memorization_hidden = extract_group(matched_memorization, memorization_prompt_key)
    print("Matched hidden-state extraction complete")
    print("reasoning_hidden shape    :", tuple(reasoning_hidden.shape))
    print("memorization_hidden shape :", tuple(memorization_hidden.shape))
    return {
        "reasoning_hidden_states": reasoning_hidden,
        "memorization_hidden_states": memorization_hidden,
        "matched_reasoning_metadata": matched_reasoning,
        "matched_memorization_metadata": matched_memorization,
        "extraction_config": {
            "model": "Qwen3-0.6B custom implementation",
            "tokenizer": "custom Qwen3 tokenizer",
            "representation": "last_formatted_prompt_token",
            "position": -1,
            "boundary": "post_transformer_block_pre_final_norm",
            "use_kv_cache": False,
            "generation": False,
            "num_reasoning_samples": int(reasoning_hidden.shape[0]),
            "num_memorization_samples": int(memorization_hidden.shape[0]),
            "n_layers": int(reasoning_hidden.shape[1]),
            "hidden_dim": int(reasoning_hidden.shape[2]),
        },
    }
````

````python
def final_layer_matrix(extraction_result):
    """Return final-layer X and labels from extracted matched hidden states."""
    reasoning_X = extraction_result["reasoning_hidden_states"][:, -1, :]
    memorization_X = extraction_result["memorization_hidden_states"][:, -1, :]
    X = torch.cat([reasoning_X, memorization_X], dim=0).float().cpu().numpy()
    y = np.asarray([REASONING_LABEL] * reasoning_X.shape[0] + [MEMORIZATION_LABEL] * memorization_X.shape[0], dtype=int)
    return X, y, reasoning_X, memorization_X
````

````python
def _json_default(obj):
    if isinstance(obj, np.integer):
        return int(obj)
    if isinstance(obj, np.floating):
        return float(obj)
    if isinstance(obj, np.ndarray):
        return obj.tolist()
    if isinstance(obj, torch.Tensor):
        return obj.detach().cpu().tolist()
    return str(obj)
````

````python
def save_experiment(
    match_result,
    length_verification,
    length_baseline_results,
    extraction_result=None,
    pca_results=None,
    hidden_probe_results=None,
    random_label_results=None,
    comparison_table=None,
    base_dir=".",
):
    """Save metadata, tensors, and result summaries without overwriting original representations."""
    base_dir = Path(base_dir)
    paths = {
        "matched_data_dir": base_dir / "data" / "matched",
        "length_results_dir": base_dir / "results" / "length_matching",
        "pca_results_dir": base_dir / "results" / "pca",
        "probe_results_dir": base_dir / "results" / "linear_probe",
        "matched_repr_dir": base_dir / "representations" / "length_matched",
    }
    for path in paths.values():
        path.mkdir(parents=True, exist_ok=True)

    matched_dataset_path = paths["matched_data_dir"] / "math500_popqa_length_matched.json"
    with matched_dataset_path.open("w", encoding="utf-8") as f:
        json.dump(match_result["matched_dataset"], f, indent=2, ensure_ascii=False, default=_json_default)

    match_summary = {
        "matched_reasoning_count": len(match_result["matched_reasoning"]),
        "matched_memorization_count": len(match_result["matched_memorization"]),
        "length_verification": length_verification,
        "length_only_baseline": length_baseline_results,
    }
    match_summary_path = paths["length_results_dir"] / "length_matching_summary.json"
    with match_summary_path.open("w", encoding="utf-8") as f:
        json.dump(match_summary, f, indent=2, ensure_ascii=False, default=_json_default)

    saved = {
        "matched_dataset": str(matched_dataset_path),
        "length_matching_summary": str(match_summary_path),
    }

    if extraction_result is not None:
        hidden_path = paths["matched_repr_dir"] / "matched_hidden_states.pt"
        torch.save(extraction_result, hidden_path)
        saved["matched_hidden_states"] = str(hidden_path)
        X, y, reasoning_X, memorization_X = final_layer_matrix(extraction_result)
        final_layer_path = paths["matched_repr_dir"] / "matched_final_layer_representations.pt"
        torch.save(
            {
                "reasoning_X": reasoning_X,
                "memorization_X": memorization_X,
                "X": torch.tensor(X),
                "y": torch.tensor(y),
                "metadata": match_result["matched_dataset"],
            },
            final_layer_path,
        )
        saved["matched_final_layer_representations"] = str(final_layer_path)

    if pca_results is not None:
        pca_path = paths["pca_results_dir"] / "length_matched_pca_results.npz"
        np.savez(
            pca_path,
            X_pca=pca_results["X_pca"],
            pc1_variance=pca_results["pc1_variance"],
            pc2_variance=pca_results["pc2_variance"],
            pc1_pc2_variance=pca_results["pc1_pc2_variance"],
        )
        saved["pca_results"] = str(pca_path)

    probe_summary = {"hidden_probe_results": hidden_probe_results, "random_label_results": random_label_results}
    probe_summary_path = paths["probe_results_dir"] / "length_matched_probe_summary.json"
    with probe_summary_path.open("w", encoding="utf-8") as f:
        json.dump(probe_summary, f, indent=2, ensure_ascii=False, default=_json_default)
    saved["probe_summary"] = str(probe_summary_path)

    if comparison_table is not None:
        comparison_path = paths["length_results_dir"] / "original_vs_length_matched.csv"
        comparison_table.to_csv(comparison_path, index=False)
        saved["comparison_table"] = str(comparison_path)

    print("\nSaved experiment artifacts")
    print("=" * 32)
    for name, path in saved.items():
        print(f"{name:38}: {path}")
    return saved
````

````python
def print_experiment_summary(
    match_result,
    length_verification,
    length_baseline_results,
    pca_results=None,
    hidden_probe_results=None,
    random_label_results=None,
    saved_paths=None,
):
    """Print the final concise report for this controlled experiment."""
    print("\nConcise length-matched experiment summary")
    print("=" * 48)
    print(
        "matched dataset size:",
        len(match_result["matched_reasoning"]),
        "reasoning +",
        len(match_result["matched_memorization"]),
        "memorization",
    )
    print("\nAfter-match length statistics")
    print(
        "reasoning mean/median/std:",
        f"{length_verification['reasoning_stats']['mean']:.3f}",
        f"{length_verification['reasoning_stats']['median']:.3f}",
        f"{length_verification['reasoning_stats']['std']:.3f}",
    )
    print(
        "memorization mean/median/std:",
        f"{length_verification['memorization_stats']['mean']:.3f}",
        f"{length_verification['memorization_stats']['median']:.3f}",
        f"{length_verification['memorization_stats']['std']:.3f}",
    )
    print("\nLength-only baseline")
    print(f"accuracy: {length_baseline_results['accuracy']['mean']:.4f} +/- {length_baseline_results['accuracy']['std']:.4f}")
    print(f"auroc   : {length_baseline_results['auroc']['mean']:.4f} +/- {length_baseline_results['auroc']['std']:.4f}")
    if pca_results is not None:
        print("\nPCA explained variance")
        print(f"PC1: {pca_results['pc1_variance']:.4f}")
        print(f"PC2: {pca_results['pc2_variance']:.4f}")
        print(f"PC1+PC2: {pca_results['pc1_pc2_variance']:.4f}")
    if hidden_probe_results is not None:
        print("\n5-fold hidden-state linear probe")
        print(f"accuracy: {hidden_probe_results['accuracy']['mean']:.4f} +/- {hidden_probe_results['accuracy']['std']:.4f}")
        print(f"auroc   : {hidden_probe_results['auroc']['mean']:.4f} +/- {hidden_probe_results['auroc']['std']:.4f}")
    if random_label_results is not None:
        print("\nRandom-label baseline")
        print(f"accuracy: {random_label_results['accuracy']['mean']:.4f} +/- {random_label_results['accuracy']['std']:.4f}")
        print(f"auroc   : {random_label_results['auroc']['mean']:.4f} +/- {random_label_results['auroc']['std']:.4f}")
    if saved_paths is not None:
        print("\nSaved paths")
        for name, path in saved_paths.items():
            print(f"{name}: {path}")
````
