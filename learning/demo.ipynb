```python
## ok we have to design a network, with 2 inputs, 2 hidden neurons, and 2 ouputs.


## first let's create the input 
## we will have 3 as number of training examples, with 2 inputs, 

# let's create the input with numpy first,

import numpy as np

X = np.random.rand(3,2) # shape 3,2

```

```python
b1 = np.random.rand(2)  # shape 2, 
```

```python
# now we got inpput and bias, 
# now we need weights 
W1 = np.random.rand(2,2)  # shape 2,2
```

```python
## now we have the input, we have the weights and bias,
# so let's make the calculation happen till the hiddne layer first..

Z1 = X @ W1 + b1 ## here bias will be broadcasted to every row...    ## shape 3,2 @ 2,2 + 2,  ---> 3,2   and bias  will be broascasted to each column accordingly in each row... 
A1 = np.maximum(0, Z1)   ## shape 3,2

#so this is for the layer 1 from input to layer 1
```

```python
## now from hidden layer 1 to output..
## let's frist create the weight 2
W2 = np.random.rand(2,2) # shape: 2,2
b2 = np.random.rand(2)  # shape 2,

Z2 = A1 @ W2 + b2 # shape: 3,2 @ 2,2 + 2, ---> 3,2


## here softmax comes...
Z2_shifted = Z2 - np.max(Z2, axis=1, keepdims=True)  ## shape: 3,2 - 3,2 ---> 3,1 --> 3,2  by numpy broadcast like copy the same elements  again. to make it 3,2
# shape: Z2_shifted 3,2

exp_Z2 = np.exp(Z2_shifted)  # shape: 3,2

p = exp_Z2 / np.sum(exp_Z2, axis=1, keepdims=True) # shape: 3,2



```

```python
class listnode:
    def __init__(self, val=0, next=None):
        self.val = val 
        self.next = next

        
```

```python
class linkedlist:
    def __init__(self, head=None):
        self.head = head


    def traverse(self):

        curr = self.head
        while curr is not None:
            print(curr.val)
            curr = curr.next

    def push_to_map(self):

        map = {}
        curr = self.head
        while curr is not None:
            val = curr.val
            if val in map:
                map[val].append(curr)
            else:
                map[val] = [curr]
            
            curr = curr.next

        return map
            
```

```python

a = listnode(40)
b = listnode(20)
c = listnode(30)
d = listnode(50)
e = listnode(30)


a.next = b
b.next = c
c.next = d
d.next = e

ll = linkedlist(a)

ll
```

```python
ll.traverse()
```

```python


map = ll.push_to_map()
```
